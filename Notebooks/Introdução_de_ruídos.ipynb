##Injetando Ruídos

```python
import pandas as pd
import numpy as np

# Reprodutibilidade
np.random.seed(42)


# Ruído 1: 0.5% de valores ausentes (NaN) em duas colunas pontuais
for col in ['População', 'DH_Outros Afetados']:
    idx_nan = df.sample(frac=0.005).index
    df.loc[idx_nan, col] = np.nan

# Ruído 2: 3 linhas com estado em minúsculo em 'UF'
idx_uf = df.sample(n=3).index
df.loc[idx_uf, 'UF'] = df.loc[idx_uf, 'UF'].str.lower()

# Ruído 3: 2 linhas duplicadas
duplicadas = df.sample(n=2)
df = pd.concat([df, duplicadas], ignore_index=True)

# Ruído 4: Inconsistência lógica (Número de mortos maior que a população da cidade)
idx_outlier = df.sample(n=1).index
df.loc[idx_outlier, 'DH_Mortos'] = df.loc[idx_outlier, 'População'] + 1000

# --- SALVAR O DATASET NO SERVIDOR DO COLAB ---
nome_arquivo = 'Danos_Informados.xlsx'
df.to_excel(nome_arquivo, index=False)
print("Dataset gerado com ruídos!")

# --- FAZER O DOWNLOAD PARA O COMPUTADOR ---
files.download(nome_arquivo)
```
